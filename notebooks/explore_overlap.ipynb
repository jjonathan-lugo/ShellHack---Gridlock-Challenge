# Explore overlap
Scratch space to sanity-check the distance math before it's wired into the app/backend.
Uses the same `backend.overlap` module the API imports, so results here match production.

```python
import sys, json
sys.path.insert(0, "..")

from backend.overlap.score import build_ranked_overlaps

projects = json.load(open("../data/processed/projects.json"))
desc = [p for p in projects if "Dominion" in p["utility"]]
gpc = [p for p in projects if p not in desc]
len(desc), len(gpc)
```

```python
ranked = build_ranked_overlaps(desc, gpc)
for o in ranked:
    print(f"#{o['rank']:>2}  {o['distance_mi']:>5.2f} mi  {o['tier']:<10}  {o['project_name_a']} <-> {o['project_name_b']}")
```

## Cross-check against the spreadsheet
`data/Projects_Overlaps.xlsx` has a hand/tool-built `overlaps` sheet already. This should match it exactly (same 6 overlaps, same top-ranked pair at ~4.09 mi) — if it doesn't, something changed in the source data or the tiering logic and is worth a closer look before trusting either one.

```python
import openpyxl
wb = openpyxl.load_workbook("../data/Projects_Overlaps.xlsx", data_only=True)
ws = wb["overlaps"]
headers = [c.value for c in ws[1]]
rows = [dict(zip(headers, row)) for row in ws.iter_rows(min_row=2, values_only=True) if row[0]]
for r in sorted(rows, key=lambda r: r["distance_mi"]):
    print(r["distance_mi"], r["project_name_a"], "<->", r["project_name_b"])
```
